Import Libraries

```python
import warnings
warnings.filterwarnings("ignore")

import joblib
import pandas as pd
import matplotlib.pyplot as plt
```

Load Model

```python
model = joblib.load(
    "../saved_models/best_model.pkl"
)
```

Load Preprocessor

```python
preprocessor = joblib.load(
    "../artifacts/data_transformation/preprocessor.pkl"
)
```

Load Prediction Dataset

```python
prediction_df = pd.read_csv(
    "../data/prediction.csv"
)

prediction_df.head()
```

Dataset Shape

```python
prediction_df.shape
```

Dataset Information

```python
prediction_df.info()
```

Transform Dataset

```python
X = preprocessor.transform(
    prediction_df
)
```

Generate Predictions

```python
predictions = model.predict(
    X
)

predictions
```

Prediction Probability

```python
if hasattr(model, "predict_proba"):

    probability = model.predict_proba(
        X
    )[:,1]

else:

    probability = None
```

Add Results

```python
result = prediction_df.copy()

result["Prediction"] = predictions

if probability is not None:

    result["Probability"] = probability

result.head()
```

Prediction Summary

```python
result["Prediction"].value_counts()
```

Probability Statistics

```python
if probability is not None:

    result["Probability"].describe()
```

Save Predictions

```python
result.to_csv(
    "../reports/predictions.csv",
    index=False,
)
```

Download Preview

```python
result.head(20)
```

Visualization

```python
plt.figure(figsize=(6,4))

result["Prediction"].value_counts().plot(
    kind="bar"
)

plt.title(
    "Prediction Distribution"
)

plt.xlabel(
    "Prediction"
)

plt.ylabel(
    "Count"
)

plt.show()
```

Probability Distribution

```python
if probability is not None:

    plt.figure(figsize=(7,4))

    plt.hist(
        probability,
        bins=20,
    )

    plt.title(
        "Prediction Probability Distribution"
    )

    plt.show()
```

Sample Positive Predictions

```python
result[
    result["Prediction"] == 1
].head()
```

Sample Negative Predictions

```python
result[
    result["Prediction"] == 0
].head()
```
