```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

pd.set_option("display.max_columns", None)
```

## Load Dataset

```python
df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

## Dataset Overview

```python
df.shape
```

```python
df.info()
```

```python
df.describe(include="all").T
```

## Target Variable Distribution

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Churn"
)

plt.title("Target Distribution")

plt.show()
```

```python
df["Churn"].value_counts(normalize=True) * 100
```

## Numerical Features Distribution

```python
numerical_columns = df.select_dtypes(
    include=["int64","float64"]
).columns

numerical_columns
```

```python
for column in numerical_columns:

    plt.figure(figsize=(6,4))

    sns.histplot(
        df[column],
        kde=True
    )

    plt.title(column)

    plt.show()
```

## Categorical Features Distribution

```python
categorical_columns = df.select_dtypes(
    include="object"
).columns

categorical_columns
```

```python
for column in categorical_columns:

    plt.figure(figsize=(8,4))

    sns.countplot(
        data=df,
        x=column
    )

    plt.xticks(rotation=45)

    plt.title(column)

    plt.show()
```

## Correlation Heatmap

```python
numeric_df = df.select_dtypes(
    include=["int64","float64"]
)

plt.figure(figsize=(10,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

## Boxplots

```python
for column in numerical_columns:

    plt.figure(figsize=(6,4))

    sns.boxplot(
        x=df[column]
    )

    plt.title(column)

    plt.show()
```

## Monthly Charges vs Churn

```python
sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.show()
```

## Tenure vs Churn

```python
sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

plt.show()
```

## Contract Type vs Churn

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Contract",
    hue="Churn"
)

plt.show()
```

## Internet Service vs Churn

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="InternetService",
    hue="Churn"
)

plt.show()
```

## Payment Method vs Churn

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="PaymentMethod",
    hue="Churn"
)

plt.xticks(rotation=20)

plt.show()
```

# 📌 Key Insights

- Customers with shorter tenure show higher churn.
- Higher monthly charges are associated with increased churn.
- Month-to-month contracts exhibit the highest churn rates.
- Electronic check users tend to churn more frequently.
- Fiber optic internet customers have relatively higher churn.
- The dataset shows moderate class imbalance.
